```python
import inspect
import requests
from pprint import pprint
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'


s = requests.Session()
r = s.request('get', 'https://api.github.com/events')
r.raw.read()
r.content
r.text
r.json()


def check_object(x):
    print('===============================================================================')
    print()
    pprint(type(x).mro())
    print()
    print(repr(x))
    print()


def check_func(x):
    print('===============================================================================')
    print()
    print('-> ', x.__qualname__, ': ', type(x), sep='')
    print()
    print(inspect.getfile(x))
    print()
    print(inspect.getsource(x))
    print()
```

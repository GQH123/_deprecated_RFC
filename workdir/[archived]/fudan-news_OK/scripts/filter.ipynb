```python
import os
import json
from tqdm import tqdm


def read_all_news(path):
    return {file: json.load(open(os.path.join(path, file))) for file in tqdm(os.listdir(path)) if file.endswith('.json')}


all_news = read_all_news('news/')
```

```python
def check_category(all_news):
    category_counter = {}

    for news in all_news.values():
        category = news['category']
        if category not in category_counter:
            category_counter[category] = 0
        category_counter[category] += 1
        
    print(json.dumps(category_counter, indent=4, ensure_ascii=False))
    return category_counter


check_category(all_news);
```

```python
def filter_by_length(all_news, min_length=500):  # filter out news with less than 500 characters, approximately 190 news
    return {name: news for name, news in all_news.items() if len(news['body']) >= min_length}


def filter_by_category(all_news, filtered_categories=["", "з§‘з”иҝӣеұ•"]):
    return {name: news for name, news in all_news.items() if news['category'] not in filtered_categories}


def filter_all(all_news):
    return filter_by_category(filter_by_length(all_news))


all_news = filter_all(all_news)
```

```python
os.makedirs('filtered_news', exist_ok=True)
for name, news in tqdm(all_news.items()):
    json.dump(news, open(os.path.join('filtered_news', f'{name}'), 'w'), indent=4, ensure_ascii=False)
```

```python
import os

def setting_up_proxy(proxy=None, proxy_type='http', verbose=True):
    supported_proxy_types = ['http', 'https', 'socks4', 'socks5', 'all']
    assert proxy_type in supported_proxy_types, f"proxy type {repr(proxy_type)} not supported, only support {supported_proxy_types}"
    if proxy is None:
        proxy = os.environ.get(f'{proxy_type}_proxy')
    if proxy is None:
        return
    if verbose:
        print(f'setting up proxy {repr(proxy)} for {repr(proxy_type)}')
    os.environ[f'{proxy_type}_proxy'] = proxy


default_proxy_config = {
    'http': 'http://10.176.52.116:7890',
    'https': 'http://10.176.52.116:7890',
    'all': 'socks5://10.176.52.116:7890',
}


def setting_up_proxy_from_config(proxy_config=default_proxy_config, verbose=True):
    for proxy_type, proxy_url in proxy_config.items():
        setting_up_proxy(proxy=proxy_url, proxy_type=proxy_type, verbose=verbose)
    print()


# setting_up_proxy_from_config()
```

```python
import json

from RFC.utils.parse import (
        BeautifulSoup,
        parse,
    )

parse_config = {
    'body': {
        ('attr', 'div', 'class', 'wp_articlecontent', None): {
            ('result', 'text', None): {},
        },
    },
    'category': {
        ('attr', 'span', 'class', 'artCat_v1', None): {
            ('result', 'text', None): {},
        },
    },
    'title': {
        ('attr', 'h1', 'class', 'arti_title', None): {
            ('result', 'text', None): {},
        },
    },
    'metas': {
        ('attr', 'p', 'class', 'arti_metas', 0): {
            ('type', 'span', None): {},
            # ('result', 'text', None): {},
        },
    },
}

result = {}
result['title'] = parse(BeautifulSoup(resp.content), parse_config['title'], return_str=True, debug=False)[0]
result['category'] = parse(BeautifulSoup(resp.content), parse_config['category'], return_str=True, debug=False)[0]
result['metas'] = {}
metas = parse(BeautifulSoup(resp.content), parse_config['metas'], return_str=False, debug=False)
for meta in metas:
    if '：' not in meta:
        continue
    k, v = meta.split('：', 1)
    result['metas'][k] = v
result['body'] = parse(BeautifulSoup(resp.content), parse_config['body'], return_str=True, debug=False)[0]
print(json.dumps(result, indent=4, ensure_ascii=False))
```

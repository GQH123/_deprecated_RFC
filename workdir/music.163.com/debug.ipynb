```python
import os

def setting_up_proxy(proxy=None, proxy_type='http', verbose=True):
    supported_proxy_types = ['http', 'https', 'socks4', 'socks5', 'all']
    assert proxy_type in supported_proxy_types, f"proxy type {repr(proxy_type)} not supported, only support {supported_proxy_types}"
    if proxy is None:
        proxy = os.environ.get(f'{proxy_type}_proxy')
    if proxy is None:
        return
    if verbose:
        print(f'setting up proxy {repr(proxy)} for {repr(proxy_type)}')
    os.environ[f'{proxy_type}_proxy'] = proxy


default_proxy_config = {
    'http': 'http://127.0.0.1:7890',
    'https': 'http://127.0.0.1:7890',
    'all': 'socks5://127.0.0.1:7890',
}


# default_proxy_config = {
#     'http': 'http://10.176.52.116:7890',
#     'https': 'http://10.176.52.116:7890',
#     'all': 'socks5://10.176.52.116:7890',
# }


def setting_up_proxy_from_config(proxy_config=default_proxy_config, verbose=True):
    for proxy_type, proxy_url in proxy_config.items():
        setting_up_proxy(proxy=proxy_url, proxy_type=proxy_type, verbose=verbose)
    print()


# setting_up_proxy_from_config()
```

```python
netease_host = "https://music.163.com"
api_store = {
    'url': "/api/song/enhance/player/url/v1",  # https://github.com/XiaoMengXinX/Music163Api-Go/blob/master/api/songURL.go
    'lyrics': "/api/song/lyric",  # https://github.com/XiaoMengXinX/Music163Api-Go/blob/master/api/songLyric.go
    'download_url': "/api/song/enhance/download/url",  # https://github.com/XiaoMengXinX/Music163Api-Go/blob/master/api/songDownloadURL.go
    'song_details': "/api/v3/song/detail",  # https://github.com/XiaoMengXinX/Music163Api-Go/blob/master/api/songDetail.go
    'short_url': "/api/middle/shorturl/generate",  # https://github.com/XiaoMengXinX/Music163Api-Go/blob/master/api/shortURL.go
    'album_details': "/api/album/v3/detail",  # https://github.com/XiaoMengXinX/Music163Api-Go/blob/master/api/albumDetail.go
}


def get_api(type):
    if type not in api_store:
        raise ValueError(f"api type {type} not found, available types are {api_store.keys()}")
    return netease_host + api_store[type]
```
